```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
df = pd.read_csv("../data/processed/english_hindi.csv")
```

```python
X = df['clean_text'] #input
y = df['labels'] #target
```

```python
print(X.shape)
print(y.shape)
```

```python
"""converting ham, spam into numbers because machine learning classifiers generall work with numeric target labels"""
```

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
print(label_encoder.classes_)
print(y_encoded[:20])
```

```python
"""this dataset is imbalanced, pay importance to stratification"""
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded,test_size=0.20, random_state=42, stratify=y_encoded)
""" this gives us approximately 80% training 20% testing stratify ensures that spam/ham ration remains approx the same in both the sets"""
```

```python
"""verify the split"""
print("training:", len(X_train))
print("testing:", len(X_test))

print("Training distribution:")
print(pd.Series(y_train).value_counts())

print("Testing distribution:")
print(pd.Series(y_test).value_counts())
```

tf-idf

```python
tfidf = TfidfVectorizer(ngram_range=(1,2), min_df=2, max_df=0.95) # unigrams and bigrams
X_train_tfidf = tfidf.fit_transform(X_train) #vectorizer needs to learn its vocab and idf statistics from the training data only 
X_test_tfidf = tfidf.transform(X_test)
# min_df ignore words appearning in only one document
# max_df ignore terms appearing in more than 95% of the odcument
```

```python
print("X_test shape:", X_test_tfidf.shape)
print("X_test shape:" , X_test_tfidf.shape)
```
